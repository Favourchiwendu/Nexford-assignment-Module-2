```python
# using panda t6o read the csv file 

import pandas as pd
df=pd.read_csv("Total.csv") 
df.head()
```

```python
#defining the employee function

def employee_function(name):
    message = df[df["EmployeeName"]==name]
    return message
    
```

```python
#practice to test the function with an employee name 

print(employee_function("PATRICK GARDNER"))
```

```python
#modifying the function to take any case used to type employee name

def employee_function(name):
    message = df[df['EmployeeName'].str.lower() == name.lower()]
    return message
    
```

```python
#practice to test the function with an employee name scattered 


print(employee_function("paTRIck gardner"))
```

```python
#modify the function to create the data dictionary

def employee_function(name):
    message = df[df['EmployeeName'].str.lower() == name.lower()]
    return message.iloc[0].to_dict()
```

```python
#practice to test the data dictionary with an employee name

print(employee_function("christopher chong"))
```

```python
#error handling in the function

def employee_function(name):
    try:
        message = df[df['EmployeeName'].str.lower() == name.lower()]
        if message.empty:
            return {'error': 'Employee not found'}
        return message.iloc[0].to_dict()
    except Exception as e:
        return {'error': str(e)}
```

```python
print(employee_function("christopher favour"))
```

```python
#importing zipfile

import zipfile
```

```python
result = employee_function('PATRICK GARDNER')
# Save result to CSV
csv_filename = 'Employee_Profile.csv' 
pd.DataFrame([result]).to_csv(csv_filename, index=False)

# Create a ZIP file and add the CSV
zip_filename = 'Employee_Profile.zip'
with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
    zipf.write(csv_filename)



print(f"Saved employee details in zipped file: {zip_filename}")
```
